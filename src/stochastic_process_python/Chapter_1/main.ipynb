Financial Instruments are asssets that can be traded, or they can also be seen as pacakges of capital that may be treaded

Assets can be in the form of cash or another financial instrument. 

The term security refers to a fungible, negotioable financial instrumnet that holds some monetary value.

Securitys can be broadly classified into the following:

- Debt: These are risk-free instrument and "secure" in the true sense, such as bank deposits and bonds. You keep getting a fixed interest as a return throughout
  the period once you make a fixed deposit in a bank.
- Equity: Company shares as stocks
- Derivatives: It is a delayed delivery contract agreement between two parties whoes value depened on another instrument. Exampls of this are options, futures, swaps, 
  etc. In this book, we will be discussing options only.

Finance is about allocating funds for business with an assessment of risk and profits.

You can make a profit just by putting a bulk amount of money as fixed deposits in any public sector bank that always gives you a fixed return, or else you may choose to put it in the share market that may give you a higher return but with chances of suddent loss.

It is challenging to compute stock price, unlike simple interest formula studied in the 10th standard. Stock prices may not follow a known deterministic pattern.

# Why Stochastic

Stochastic is a technical word associated with random probability distributions and non-deterministic. 

It means that something is there whose value(s) cannot be determined precisely, and there are some uncertainties around it. stock values are potential examples
of such things. By determined we mean forecasting future values to help you plan your investments.

Knowing the future values of stocks or securities involves fair amount of mathematical modeling. Stock values depend on time and resemble a stream.

That vies the idea of modeling them with stochastic process. 

Stochastic process is a stream fo random values without a particular state.
