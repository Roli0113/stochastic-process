### Options

Futures and options are one of the most popular derivates traded in a regulated exchange. 

Formally, a derivative is a financial instrumment whos value depends on other, more basic, underlying assets. 

We will discuss stock options, which is a derivate that depends on the underlying stock price. 

In options a contract is what being sold between the buyer and the seller.

This contract is formally known as the options, which can be bought by the option buyer with a paid premium. 

THis allows the buyer to exercise their right to buy or sell when the underlying stock price moves in their favor. 

THis is a right not an obligation if the buyer choose to do that.

A ***call*** option gives the option buyer the right to buy the underlying asset by a certain date for a certain price

A ***put*** option gives the option seller the right to sell the underlying asset by a certain date for a certain price.

The price of the contract is known as the exercise price or the strike price, and the exercise date is known as the experitaion date or maturity.

The payoff if the value of an option, which is the difference between strike price - stock price at the time, assuming that paid premium will not return. 

The statistical core problem is estimating the value of the option premium that you should pay to buy an option that will hedge against a potential loss. 

As the function of the underlying assets and time, options are a stochastic process, a function of another stochastic process, maing it a complex problem to solve.

***Two methods to solve option Premium Evaluation Problem***
- Risk- Neutral Probabilistic Method
- Stochastic Partial Differential Equation (SPDE) which is based on the Black-Scholes Model.
